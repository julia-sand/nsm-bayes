# Loading Simulation Data

This notebook demonstrates how to load the `.pkl` files generated by the simulation scripts (`run_sir.py` and `run_gnk.py`).

```python
import pickle
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# Update this path to the directory where your simulation data is stored
data_dir = Path("data/sir")

def load_pkl_files(directory, pattern="*.pkl"):
    """Loads all pickle files in a directory matching the pattern."""
    data = {}
    for pkl_path in directory.glob(pattern):
        with open(pkl_path, "rb") as f:
            data[pkl_path.name] = pickle.load(f)
    return data

try:
    files_data = load_pkl_files(data_dir)
    print(f"Successfully loaded {len(files_data)} files.")
    print("Files found:", list(files_data.keys()))
except FileNotFoundError:
    print(f"Error: The directory {data_dir} was not found. Please update the `data_dir` path.")
```

## Exploring the Data

You can now access the loaded data using the filename as the key.

```python
# Example: Accessing a specific file (replace with an actual filename from your output)
if files_data:
    first_file = list(files_data.keys())[0]
    content = files_data[first_file]
    print(f"Content of {first_file}:")
    print(content)
else:
    print("No data loaded to explore.")
```
